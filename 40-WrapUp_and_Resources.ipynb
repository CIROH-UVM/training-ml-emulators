# **Workshop Wrap-Up and Resources**
## **Machine-Learned Emulators for Teleconnection Discovery and Uncertainty Quantification**
In this workshop, we explored how machine-learned emulators can be used to approximate expensive simulation-based environmental models and make their outputs easier to evaluate, interpret, and communicate.
## 1. **What We Covered**
We walked through a complete emulator-analysis workflow:
1. **Problem formulation**
   - Why expensive process-based CHANS simulations motivate the use of machine-learned emulators
2. **Dataset and Model setup**
   - How simulator-derived tabular datasets are organized into input drivers and output targets
   - How Random Forest, LightGBM, and XGBoost can be used as emulator models
3. **Training and Evaluation**
   - How to split data into training, validation, and test sets
   - How to train regression and classification emulators
   - How to evaluate regression models using MAE, RMSE, Bias, and $R^2$
   - How to evaluate classification models using Precision, Recall, F1-score, and confusion metrics
4. **Feature importance**
   - How to identify which variables the trained emulator relies on most
5. **PDPs and Bootstrap PDPs** 
   - How to examine how predictions change as important variables vary
   - How to assess whether PDP patterns are stable or uncertain under resampling


## **2. How This Workflow Transfers to Other Fields**

The workflow used in this notebook is **not limited to chlorophyll-a or waterquality modeling**.
The same emulator-analysis pipeline can be **reused for many simulation based environmental problems**, such as:

- Flood simulator emulation
- Drought-risk surrogate modeling
- Crop-yield scenario exploration
- Lake and reservoir water-quality analysis
- Climate-impact scenario screening
- Policy and management intervention analysis
- National Water Model evaluation workflows
- Hydrologic forecast post-processing
- Coupled human-natural systems modeling

The only requirement is that the simulator or modeling system can produce a structured input-output dataset.

## **3. How to Use This Workflow in Your Own Project**

To adapt this workflow to your own research or operational problem, ask the following questions:

### **Step 1: What simulator or model are you trying to emulate?**

Examples:

- Hydrologic model
- Water-quality model
- Climate-impact model
- Crop model
- Flood-risk model
- Reservoir operations model
- Coupled human-natural systems model

### **Step 2: What are the inputs?**

May include:

- Climate variables
- Hydrologic variables
- Land-use variables
- Soil or terrain variables
- Management/Policy scenarios
- Boundary conditions

### **Step 3: What are the outputs?**

Outputs may be continuous or categorical.

Examples:

- Streamflow magnitude
- Flood depth
- Chlorophyll-a concentration
- Trophic-state class
- Drought-risk category
- Crop yield
- Reservoir storage

### **Step 4: What type of emulator is needed?**

- Use **regression** if the target is continuous
- Use **classification** if the target is categorical
- Use both if you need complementary views of the system

### **Step 5: Lastly, but most important one: How will you interpret the emulator?**

Useful interpretation tools include:
- Feature importance
- PDPs
- Bootstrap PDPs

## **4. Interpretation Guidelines**

Emulator interpretation should be careful and scientifically grounded.

### **Feature Importance**

Feature importance helps answer:

**Which variables were most useful for prediction inside the trained emulator?**

However:

- High importance does not automatically imply causality
- Importance values may differ across model families
- Interpretation depends on the data used to train the emulator

### **Partial Dependence Plots**

PDPs help answer:

**How does the trained emulator's prediction change as one feature varies?**

PDPs can reveal:

- Positive or negative associations
- Nonlinear responses
- Threshold-like behavior
- Class-specific response patterns

However:

- PDPs describe model behavior, not direct real-world causality
- PDPs may evaluate feature combinations that are rare or unrealistic
- PDPs should be interpreted alongside domain knowledge

### **Bootstrap PDPs**

Bootstrap PDPs help assess whether PDP patterns are stable.

If the bootstrap curves are similar and the uncertainty bands are narrow, the response pattern is more stable.

If the curves vary widely or the uncertainty bands are broad, the relationship should be interpreted cautiously.

## **5. Final Cautions**

Important cautions:

- The emulator is trained on simulator outputs, not direct reality
- Inherits assumptions, simplifications, and biases from the original simulator
- Good predictive performance does not prove causal understanding
- Feature importance identifies predictive relevance, not causal mechanisms
- PDPs show emulator-learned response behavior, not direct physical laws
- Bootstrap PDPs improve stability assessment, but they do not create causal certainty
- Emulator results should be used as evidence for further investigation, not as final proof

The safest interpretation is:

**The emulator helps us understand how the simulator behaves under different input conditions**

## **6. Key Takeaways**

- **Machine-learned emulators** can approximate **expensive simulator behavior** at a much **lower computational cost**
- **Regression and classification** provide **complementary views** of environmental system response
- **Feature importance** helps identify variables that are **predictive inside the trained emulator**
- **PDPs** show how emulator predictions change as **important variables vary**
- **Bootstrap PDPs** help distinguish **stable response patterns** from **uncertain or less reliable patterns**
- Emulator interpretation can support **teleconnection discovery** by highlighting **distant or indirect drivers** associated with local outcomes
- **Strong predictive importance does not automatically imply real-world causality**
- This workflow is **transferable** to many **simulation-based environmental and water-resources applications**
