## **[Machine-Learned Emulator Workflow for Teleconnection Discovery and Uncertainty Quantification](https://www.mdpi.com/2073-4441/18/1/79)**

Welcome to this DevCon workshop on machine-learned emulators for coupled socio-hydrological systems. In this workshop, we will explore how machine learning models can be used as fast surrogate models for expensive process-based environmental simulations.

The main goal is not only to make predictions, but to understand how trained emulators can help us investigate system behavior, identify important drivers, quantify uncertainty, and interpret possible teleconnection-like patterns across human and natural systems.

We will use a coupled human-natural systems modeling application focused on water quality in Missisquoi Bay of Lake Champlain. The workflow demonstrates how emulator models can support scientific interpretation of harmful algal bloom dynamics, trophic-state shifts, and the relative influence of hydro-climatic, nutrient-management, and in-lake biogeochemical processes.

## **1. What You Will Learn**

By the end of this workshop, you will be able to:

- Understand why machine-learned emulators are useful when full environmental simulations are too expensive to run repeatedly.
- Explain how an emulator can act as a fast approximation of a complex coupled human-natural systems model.
- Understand how watershed, climate, nutrient, and management conditions can be connected to lake water-quality outcomes.
- Describe the idea of teleconnection in simple terms: distant or indirect processes influencing local environmental responses.
- Explore how chlorophyll-a concentration and trophic-state categories are used to describe water-quality conditions in the lake.
- See how emulator models can be used to study both continuous water-quality responses and ecological regime shifts.
- Understand which environmental and management conditions have the strongest influence on model predictions.
- Learn how feature importance helps identify the drivers most strongly linked to chlorophyll-a and trophic-state outcomes.
- Learn how Partial Dependence Plots help us examine what happens to model predictions when one important condition changes.
- Recognize that emulator results should be interpreted as learned relationships within the simulation system, not automatically as real-world causal proof.
- Understand how this workflow can be transferred to other water resources, environmental modeling, and coupled human-natural systems problems.
